# Labels analysis

The labels from TransTVDiag are more imbalanced than the labels from TVDiag.


```python
import numpy as np
import pandas as pd
```

```python
tvdiag_labels_df = pd.read_csv('data/gaia/gaia.csv')
tvdiag_labels_df.head()
```

```python
print(tvdiag_labels_df[tvdiag_labels_df['data_type'] == 'train'].shape)
print(tvdiag_labels_df[tvdiag_labels_df['data_type'] == 'test'].shape)
print(tvdiag_labels_df[tvdiag_labels_df['data_type'] == 'train'].shape[0]/tvdiag_labels_df.shape[0])
```

```python
def value_counts_with_percentages(series, dropna=False, decimals=2):
    counts = series.value_counts(dropna=dropna)
    percents = (series.value_counts(normalize=True, dropna=dropna) * 100).round(decimals)
    return pd.DataFrame({
        'count': counts,
        'percent': percents
    })
# tvdiag_labels_df['anomaly_type'].value_counts()
value_counts_with_percentages(tvdiag_labels_df['anomaly_type'])
```

```python
value_counts_with_percentages(tvdiag_labels_df['instance'])
```

```python
transtvdiag_labels_df = pd.read_csv('data/gaia/label_15_85.csv')
transtvdiag_labels_df.head()
```

```python
value_counts_with_percentages(transtvdiag_labels_df['anomaly_type'])
```

```python
value_counts_with_percentages(transtvdiag_labels_df['instance'])
```
