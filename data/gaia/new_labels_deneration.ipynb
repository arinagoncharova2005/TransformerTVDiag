```python
def value_counts_with_percentages(series, dropna=False, decimals=2):
    counts = series.value_counts(dropna=dropna)
    percents = (series.value_counts(normalize=True, dropna=dropna) * 100).round(decimals)
    return pd.DataFrame({
        'count': counts,
        'percent': percents
    })
# tvdiag_labels_df['anomaly_type'].value_counts()
value_counts_with_percentages(labels_without_login_failure_df['anomaly_type'])
```

```python
print('Train labels distribution without login failure')
value_counts_with_percentages(labels_without_login_failure_df[labels_without_login_failure_df['data_type'] == 'train']['anomaly_type'])
```

```python
print('Test labels distribution without login failure')
value_counts_with_percentages(labels_without_login_failure_df[labels_without_login_failure_df['data_type'] == 'test']['anomaly_type'])
```

```python
print('Train labels distribution for full dataset')
value_counts_with_percentages(full_labels_df[full_labels_df['data_type'] == 'train']['anomaly_type'])
```
